```python
import matplotlib.pyplot as plt
```

```python
x_val  = [a for a in range(10) if a%2==0]
y_val  = [b for b in range(10) if b%2 != 0]
```

```python
x_val,y_val
```

```python
plt.plot(x_val,y_val)
plt.title('Demo')
plt.xlabel('X_val')
```
